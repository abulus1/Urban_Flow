````python
df = pd.read_csv("urban_flow/data/raw/speeding_fines.csv")

def normalizar_fecha(fecha: str) -> str:
  """Normaliza la fecha al formato YYYY-MM-DD."""

  for fmt in ("%d/%m/%Y", "%Y/%m/%d", "%Y-%m-%d"):
    try:
      return pd.to_datetime(fecha, format=fmt).strftime("%Y-%m-%d")
    except:
      pass
  return "1932-01-01"

df["fecha"] = df["fecha"].apply(normalizar_fecha)
print(df["fecha"].head(10))
````

````python
def normalizar_hora(hora: str) -> str:
  """Normaliza la hora al formato HH:MM de 24hs."""

  for fmt in ("%I:%M %p", "%H:%M"):
    try:
      return pd.to_datetime(hora, format=fmt).strftime("%H:%M")
    except:
      pass
  return "00:00"

df["hora"] = df["hora"].apply(normalizar_hora)
print(df["hora"].head(10))
````

````python
def normalizar_ubicacion(ubicacion: str) -> str:
  """Quita caracteres especiales y convierte a mayúsculas."""

  return re.sub(r"[^a-zA-Z0-9\s]", "", ubicacion).upper().strip()

df["ubicacion"] = df["ubicacion"].apply(normalizar_ubicacion)
print(df["ubicacion"].head(10))
````

````python
def normalizar_patente(patente):
  """Limpia la patente quitando caracteres especiales."""

  if pd.isna(patente):
    return pd.NA
  limpia = re.sub(r"[^a-zA-Z0-9]", "", str(patente)).upper()
  return limpia if limpia else pd.NA

df["patente"] = df["patente"].apply(normalizar_patente)
print(df["patente"].tail(10))
````

````python
columnas_relevantes = [
  "velocidad_registrada",
  "velocidad_maxima",
  "patente",
]

antes = len(df)
df = df.dropna(subset=columnas_relevantes)
eliminadas = antes - len(df)

print(f"Se eliminaron {eliminadas} filas.")
print("Las columnas observadas son:")
for col in columnas_relevantes:
  print(f" - {col}")
````

````python
columnas_outliers = ["velocidad_registrada", "velocidad_maxima"]

antes = len(df)
for col in columnas_outliers:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  df = df[
    (df[col] >= q1 - 1.5 * iqr) &
    (df[col] <= q3 + 1.5 * iqr)
  ]

eliminadas = antes - len(df)
print(f"Se eliminaron {eliminadas} filas.")
print("Las columnas observadas son:")
for col in columnas_outliers:
  print(f" - {col}")
````

````python
df["exceso_velocidad_real"] = (
  df["velocidad_registrada"] - df["velocidad_maxima"]
)
print(df[["patente", "exceso_velocidad_real"]].head(10))
````

````python
df["exceso_velocidad"] = (
  df["velocidad_registrada"] -
  df["velocidad_maxima"] -
  df["velocidad_maxima"] * 0.05
)
print(df[["patente", "exceso_velocidad"]].head(10))
````

````python
antes = len(df)
df = df[df["exceso_velocidad"] > 0]
eliminadas = antes - len(df)
print(f"Se eliminaron {eliminadas} filas.")
````

## Ejercicio 04

> Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
class FineAnalyzer:
  """Analizador de multas por exceso de velocidad."""

  def __init__(self, df: pd.DataFrame) -> None:
    self.df = df.copy()

  def ranking_patentes(self) -> pd.DataFrame:
    """Retorna el top 5 de patentes más multadas."""
    resultado = (
      self.df["patente"]
      .value_counts()
      .head(5)
      .reset_index()
    )
    resultado.columns = ["patente", "cantidad"]
    resultado.index = pd.RangeIndex(start=1, stop=len(resultado) + 1)
    return resultado

  def ranking_horarios(self) -> pd.DataFrame:
    """Retorna el top 5 de horarios con más multas."""
    resultado = (
      self.df["hora"]
      .value_counts()
      .head(5)
      .reset_index()
    )
    resultado.columns = ["hora", "cantidad"]
    resultado.index = pd.RangeIndex(start=1, stop=len(resultado) + 1)
    return resultado

  def exceso_promedio(self) -> float:
    """Retorna el exceso promedio de velocidad."""
    return round(self.df["exceso_velocidad"].mean(), 2)

  def exceso_real_promedio(self) -> float:
    """Retorna el exceso real promedio de velocidad."""
    return round(self.df["exceso_velocidad_real"].mean(), 2)

  def multas_por_ubicacion(self) -> pd.DataFrame:
    """Retorna la cantidad de multas por ubicación."""
    resultado = (
      self.df.groupby("ubicacion")
      .size()
      .reset_index(name="cantidad")
      .sort_values("ubicacion")
    )
    return resultado
````
