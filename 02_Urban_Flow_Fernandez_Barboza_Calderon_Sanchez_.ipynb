En este punto realizaremos verificaciones sobre el dataset de imágenes:

- Listar todas las imágenes disponibles (no mostrar/no imprimir). Se debe mostrar el nombre y su tamaño en kb.

- Separar las imágenes en 2 grupos,"completes" y "plates" y contar cuantas imágenes hay en cada grupo. Ayuda calcular la resolución promedio. Al trabajar con la información de las imágenes vamos a utilizar un diccionario (denominado group_images) para almacenar información imortante que podemos utilizar más adelante.

La salida es
```
El grupo 'plates' contiene XX imágenes.
El grupo 'completes' contiene XX imágenes.
```

El diccionario es (guardarlo en `ubran_flow/data/interim/group_images.json`):
```
group_images = {
  'plates': [{
    'filename': contiene solo el nombre del archivo y la extensión,
    'width': contiene el ancho en pixeles,
    'height': contiene el alto en pixeles,
    'area': cálculo del área de la imagen,
    'path': contiene el path completo de la imagen original,
    'patent': va a contener la patente
  }],
  'completes': [{

    'height': contiene el alto en pixeles,
    'area': cálculo del área de la imagen,
    'path': contiene el path completo de la imagen original,
    'patent': va a contener la patente
  }],

- Desarrollar un función para mostrar 8 imágenes de forma aleatorea en forma de tabla de 2 columnas y 4 filas. Debe mostrar 4 imágenes de cada grupo. Es una función reutilizable. En este punto se deben mostrar las imágenes originales. Cada imágen debe tener un ancho de 10 pulgas y un alto de 8 pulgadas, se debe mostrar el nombre de la imagen.



````python
# Listar imagenes disponibles
IMGS_PATH = "urban_flow/data/raw/imgs/urban_flow_plates"
all_files = sorted(
  f for f in os.listdir(IMGS_PATH) if f.endswith(".jpg")
)
print(f"{"Archivo":<50} {"Tamaño (KB)":>12}")
print("-" * 64)
for fname in all_files:
  fpath = os.path.join(IMGS_PATH, fname)
  size_kb = os.path.getsize(fpath) / 1024
  print(f"{fname:<50} {size_kb:>11.2f}")
````

````python
# Clasificacion en grupos y diccionario group_images
group_images: dict = {"plates": [], "completes": []}
for fname in all_files:
  fpath = os.path.join(IMGS_PATH, fname)
  with Image.open(fpath) as img:
    w, h = img.size
  entry: dict = {
    "filename": fname,
    "width": w,
    "height": h,
    "area": w * h,
    "path": fpath,
    "patent": ""
  }
  grupo = "plates" if "_patente" in fname else "completes"
  group_images[grupo].append(entry)

res_plates = [e["area"] for e in group_images["plates"]]
res_completes = [e["area"] for e in group_images["completes"]]
print(
  f"Resolución promedio plates:    "
  f"{sum(res_plates)/len(res_plates):,.0f} px²"
)
print(
  f"Resolución promedio completes: "
  f"{sum(res_completes)/len(res_completes):,.0f} px²"
)
print(f"El grupo 'plates' contiene {len(group_images['plates'])} imágenes.")
print(f"El grupo 'completes' contiene {len(group_images['completes'])} imágenes.")

os.makedirs("urban_flow/data/interim", exist_ok=True)
with open("urban_flow/data/interim/group_images.json", "w") as f:
  json.dump(group_images, f, indent=2)
````

````python
# Funcion para mostrar imagenes
def mostrar_imagenes(
  group_images: dict,
  path_key: str = "path"
) -> None:
  """Muestra 8 imágenes aleatorias en grilla 2 cols x 4 filas.

  Args:
    group_images: diccionario con grupos plates y completes.
    path_key: clave del diccionario que indica la ruta de la imagen.
  """
  sample_p = random.sample(group_images["plates"], 4)
  sample_c = random.sample(group_images["completes"], 4)
  muestras = sample_p + sample_c
  fig, axes = plt.subplots(4, 2, figsize=(10, 32))
  axes = axes.flatten()
  for ax, entry in zip(axes, muestras):
    img = Image.open(entry[path_key])
    cmap = "gray" if img.mode == "L" else None
    ax.imshow(img, cmap=cmap)
    ax.set_title(entry["filename"])
    ax.axis("off")
  plt.tight_layout()
  plt.show()
````

````python
# Mostrar imagenes originales
mostrar_imagenes(group_images)
````

# Ejercicio 03

> puntos: 2

En este punto realizaremos un procesamiento de las imágenes, en cada punto se debe llamar a la función que muestra las imágenes trabajadas en el punto anterior.

- Conversión a escala de grises de las imágenes originales y almacenarlas en `urban_flow/data/interim/imgs/03_01_gray_scale/plates` y `urban_flow/data/interim/imgs/03_01_gray_scale/completes`, según la clasificación del ejercicio 2.   

- Realizar un suavizado de las imágenes en escala de grises y almacenarlas en `urban_flow/data/interim/imgs/03_02_blur` según la clasificación del ejercicio 2.

- Realizar una detección de bordes sobre las imágenes suavizadas para intentar detectar las patentes y almacenarlas en `urban_flow/data/interim/imgs/03_03_canny` según la clasificación del ejercicio 2.



*Conversion a escala de grises*

````python
BASE_GRAY = "urban_flow/data/interim/imgs/03_01_gray_scale"

for grupo in ("plates", "completes"):
  out_dir = os.path.join(BASE_GRAY, grupo)
  os.makedirs(out_dir, exist_ok=True)
  for entry in group_images[grupo]:
    img = cv2.imread(entry["path"])
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    out_path = os.path.join(out_dir, entry["filename"])
    cv2.imwrite(out_path, gray)
    entry["path_gray"] = out_path

mostrar_imagenes(group_images, path_key="path_gray")
````

*Suavizado de las imagenes*

````python
BASE_BLUR = "urban_flow/data/interim/imgs/03_02_blur"

for grupo in ("plates", "completes"):
  out_dir = os.path.join(BASE_BLUR, grupo)
  os.makedirs(out_dir, exist_ok=True)
  for entry in group_images[grupo]:
    gray = cv2.imread(entry["path_gray"], cv2.IMREAD_GRAYSCALE)
    blur = cv2.GaussianBlur(gray, (5, 5), 0)
    out_path = os.path.join(out_dir, entry["filename"])
    cv2.imwrite(out_path, blur)
    entry["path_blur"] = out_path

mostrar_imagenes(group_images, path_key="path_blur")
````

*Deteccion de bordes*

````python
BASE_CANNY = "urban_flow/data/interim/imgs/03_03_canny"

for grupo in ("plates", "completes"):
  out_dir = os.path.join(BASE_CANNY, grupo)
  os.makedirs(out_dir, exist_ok=True)
  for entry in group_images[grupo]:
    blur = cv2.imread(entry["path_blur"], cv2.IMREAD_GRAYSCALE)
    canny = cv2.Canny(blur, 50, 150)
    out_path = os.path.join(out_dir, entry["filename"])
    cv2.imwrite(out_path, canny)
    entry["path_canny"] = out_path

with open("urban_flow/data/interim/group_images.json", "w") as f:
  json.dump(group_images, f, indent=2)

mostrar_imagenes(group_images, path_key="path_canny")
````

Una vez extraidas las patentes se debe relacionar lo extraido con el dataset previamente tratado y almacenado en `urban_flow/data/interim/speeding_fines.csv`, para finalmente guardar el dataset en `urban_flow/data/processed/speeding_fines_image.csv`.

Para considerarlo como coincidencia se debe hacer un match de letras en orden de izquierda a derecha y el match debe ser del 80% o más.

En caso de match se deben agregar las siguientes columnas al dataframe:

Ejemplos:
| Patente Original | Patente Detectada | Match |

`IMPORTANTE: Esta es una aproximación base y sin complicaciones.` Se aceptan otras exploraciones siempre y cuando los resultados sean aceptables. Una base un poco más compleja puede ser:
 - Cierre Morfológico Rectangular: Usar un kernel rectangular horizontal para fusionar el contenido de la patente.
 - Encontrar Contornos: Detectar las formas blancas resultantes del paso anterior.
 - Filtrado Geométrico: Descartar los contornos que no parezcan una patente basándonos en:
    - Área: Ni muy chica (ruido) ni muy grande (todo el auto).
    - Relación de Aspecto (Aspect Ratio): Una patente es siempre más ancha que alta. La relación ancho / alto suele estar entre 2 y 5.
    - Forma: Intentar aproximar el contorno a un polígono de 4 lados.

`No se otorgan puntos extras por una exploración más compleja, pero si será tomado como nota de concepto para la puntuación final del cuatrimestre`

````python
#@title Instalación de [easyocr](https://github.com/JaidedAI/EasyOCR)
!pip install -q easyocr
````

````python
#@title Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

````python
from IPython.display import display, HTML, Code
import os

display(HTML("""
<style>
  .jp-RenderedText pre, .jp-RenderedHTMLCommon pre, .p-Widget pre {
    background-color: #1e1e1e !important;
    color: #d4d4d4 !important;
    border: 1px solid #FFF !important;
  }
  .highlight  { background: #1e1e1e !important; }
  .highlight pre { border: 2px solid !important; }
</style>
"""))
!rm -rf /content/sample_data .config
def visualizar_archivos_dark(directorio):
    for raiz, _, archivos in os.walk(directorio):
      if raiz.lower().endswith(('.git','.dvc', '__pycache__')):
        continue
      for nombre in archivos:
        if nombre.lower().endswith(('.py')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          display(Code(contenido, language=nombre.split('.')[-1]))
        if nombre.lower().endswith(('.md', '.csv')):
          ruta = os.path.join(raiz, nombre)
          print(f"📁 {ruta}")
          with open(ruta, 'r') as f:
              contenido = f.read()
          print("".join(contenido))
          print("\n\n")
          print("-" * 60)
          print("\n\n")

visualizar_archivos_dark('/content/')
````
